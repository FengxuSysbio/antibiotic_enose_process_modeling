# Gentamicin-derived few-shot target-domain adaptation

See the repository README and `docs/EVIDENCE_BOUNDARIES.md` before interpreting model performance.

```
from pathlib import Path
import pandas as pd
from antibiotic_enose.constants import TRANSFER_TARGETS
from antibiotic_enose.transfer.representation import GentamicinSourceRepresentation
from antibiotic_enose.transfer.evaluation import learning_curve_for_variable, summarize_learning_curves, threshold_table
```

```
DATA_DIR = Path('../data/processed/result7')
src = pd.read_csv(DATA_DIR/'Gentamicin_REAL_source.csv')
source_repr = GentamicinSourceRepresentation(n_components=6).fit(src)
```

```
all_rows=[]
for product, variables in TRANSFER_TARGETS.items():
    df=pd.read_csv(DATA_DIR/f'{product}_REAL_aligned.csv')
    for variable in variables:
        if variable in df.columns:
            all_rows.append(learning_curve_for_variable(product,df,variable,source_repr,n_repeats=100))
repeat_level=pd.concat(all_rows,ignore_index=True)
summary=summarize_learning_curves(repeat_level)
summary.head()
```

```
threshold_table(summary, threshold=0.80)
```
